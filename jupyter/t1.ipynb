```python
import os
from pathlib import Path

def count_images(directory, extensions={'.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff', '.webp'}):
    """
    统计目录及子目录中指定图片格式的文件数量
    参数：
        directory (str): 要扫描的目录路径
        extensions (set): 要统计的图片扩展名集合（不区分大小写）
    返回：
        tuple: (总数量, 各扩展名数量字典, 文件路径列表)
    """
    total = 0
    ext_counts = {ext: 0 for ext in extensions}
    all_files = []
    
    # 遍历目录树
    for root, _, files in os.walk(directory):
        for filename in files:
            file_ext = Path(filename).suffix.lower()  # 获取小写扩展名
            if file_ext in extensions:
                total += 1
                ext_counts[file_ext] += 1
                all_files.append(os.path.join(root, filename))
    
    return total, ext_counts, all_files

# 配置参数
target_dir = "/public/lab_share/nfs_field/level_1/DJI/P1/zhaoxian2022"
image_extensions = {'.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff', '.webp'}

# 执行统计
try:
    total_count, extension_counts, file_list = count_images(target_dir, image_extensions)
    
    # 打印结果报告
    from IPython.display import Markdown
    
    report = f"""
    ## 图片统计报告
    
    ​**扫描目录**: `{target_dir}`
    
    - 总图片数量: ​**{total_count}**
    - 各格式数量:
    {chr(10).join([f'    - {ext}: {count}' for ext, count in extension_counts.items()])}
    
    *前 5 个文件示例*:
    {chr(10).join([f'    - {Path(p).name}' for p in file_list[:5]])}
    """
    display(Markdown(report))
    
except FileNotFoundError:
    print(f"错误：目录不存在 {target_dir}")
except PermissionError:
    print(f"错误：没有权限访问目录 {target_dir}")
except Exception as e:
    print(f"发生未知错误: {str(e)}")
```
